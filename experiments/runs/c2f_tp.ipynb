# Experiment: C2F-TP-style diffusion (coarse-to-fine)

Re-implementation of the idea of C2F-TP (Wang et al., AAAI 2025): a coarse stage predicts a multimodal trajectory distribution from an interaction encoder, then a conditional denoising diffusion model refines it. Chosen over CDSTraj for row 9 because its two stages fit our training loop cleanly (CDSTraj's characterized diffusion module is not specified enough to reproduce). No public code was found; layer details are ours.

Part of the model comparison in `experiments/MODEL_ALTERNATIVES.md`. Run All; set `DEBUG = False` in section 2 for the real run.

| S3 | Loss = MSE + bivariate-Gaussian NLL + neighbour-goal MSE (`basic_loss`); the risk-scaled loss is used only where this notebook's M table says so (STRAP-R-based variants). A model may add its own `aux_loss` (only the diffusion model does). | Keeps one training objective; the risk-scaled loss is STRAP-R's own contribution. | Loss values are therefore not comparable across models: compare RMSE only. |
| S4 | Best checkpoint = lowest validation loss (`fit` default); test RMSE is reported at 1-5 s for that checkpoint. | Same selection rule for every model. | Selecting on val RMSE would be closer to the reported metric. |
| S5 | KNOWN LIMITATION (not fixed): positions are relative to the target at each timestep, so the target's own history is always (0, 0) and lateral speed is always 0 (NGSIM gives longitudinal speed only). The target is seen through speed, acceleration and lane features; lateral context comes from neighbours. | Kept as is (user decision) so every model gets identical inputs. | Papers feed the target's own x/y track; lane-change prediction is handicapped for all models here. |
| S6 | Neighbours = up to 15 risk-selected vehicles (S/O-field > RISK_THRESHOLD), not every nearby vehicle. | Same scene construction as STRAP. | Grid/graph papers use all vehicles in range; ours see only the risk-selected ones. |
| S7 | Output = one bivariate Gaussian per future step (mu_x, mu_y, sigma_x, sigma_y, rho); RMSE uses the mean. Multimodal models are moment-matched to one Gaussian. | The `fit` / `evaluate_trajectory` contract and how STRAP's Table I is scored. | minADE/minFDE over modes would favour multimodal models. |

**Specific to this model (M)**

| # | Decision / assumption | Why | Departure from paper / alternative |
|---|---|---|---|
| M1 | Row 9 implemented as C2F-TP-style, not CDSTraj. | Coarse + refine maps onto `fit` (main loss + `aux_loss`). | CDSTraj reports the better NGSIM number (2.85 m) but its module details are not public. |
| M2 | Coarse stage: per-vehicle FC + ELU + LSTM(64), target-query attention over real vehicles at the last step, 6 modes moment-matched to one Gaussian; trained by the shared loss (S3). | A compact interaction encoder; same output contract. | C2F-TP's spatial-temporal interaction module is larger. |
| M3 | Refinement: DDPM on the RESIDUAL (true future - coarse mean) / 5 m; 100 noise steps, betas linear 1e-4..0.2; an MLP denoiser (3 x 256, SiLU) conditioned on the encoder context and a learned step embedding predicts the noise. | Residual diffusion is how a coarse-to-fine model refines; the scale keeps residuals near unit variance. | ASSUMPTION: schedule, residual scale and denoiser size are ours. C2F-TP starts from noisy samples of the coarse distribution. |
| M4 | The denoising loss is added through the new `aux_loss` hook in `src/train.py`; the coarse mean is detached there, the context is not. | `fit` gives the true future only to `aux_loss`; the encoder learns from both objectives. | Two-stage training (coarse first, then refine) is the alternative. |
| M5 | Training-mode forward returns the coarse prediction only; eval-mode forward (validation, test, collisions) adds the refinement. | Sampling in every training step would be slow; checkpoint selection still uses the refined output. | - |
| M6 | Inference: K = 16 DDIM samples, 10 deterministic steps, fixed seed; mean = coarse mean + average residual; sigma combines coarse sigma with the samples' spread; rho from the coarse stage. | Single-Gaussian output (S7); a fixed seed makes evaluation repeatable. | Reporting minADE over the 16 samples would score the diffusion model higher. |
| M7 | The encoder runs twice per training batch (forward + aux_loss). | Keeps the model inside the `fit` contract. | Costs about 2x encoder time. |

## 4. Model

Any `torch.nn.Module` works if `forward(states, mask, dims)` returns a 3-tuple:

| output | shape | meaning |
|---|---|---|
| `traj_dist` | `(B, T_f=50, 5)` | `[mu_x, mu_y, sigma_x, sigma_y, rho]` of the target, metres relative to its last observed position (x longitudinal, y lateral); `sigma > 0`, `rho` in (-1, 1) |
| `goals` | `(B, N-1, 4)` | neighbour goals; a model without a goal head returns `states.new_zeros(B, N - 1, 4)` |
| third | anything | ignored by training/eval (STRAP returns its risk field) |

Inputs: `states (B, T_h=30, N, 11)` normalized (vehicle 0 = target, rest neighbours, zero-padded),
`mask (B, N)` True = real vehicle, `dims (B, N, 2)` raw length/width in metres.
Feature index: 0 rel-longitudinal m, 1 rel-lateral m, 2 speed/30, 3 lateral speed (always 0), 4 accel/10,
5 type/2, 6 relative lane (target - vehicle, left = +), 7 length/10, 8 width/3, 9 S-field risk, 10 O-field risk.
Raw value = `states * std + mean` with `norm_stats`. STRAP's k-means codebook is only loaded if the model has a `risk_decoder`.

**Decisions in this cell:** coarse multimodal stage (M2); residual DDPM with an MLP denoiser (M3) trained via `aux_loss` (M4); refinement only in eval mode (M5); 16 deterministic DDIM samples (M6).

```python
def moment_match(raw, weights):
    """(B, M, T, 5) raw mode outputs + (B, M) mode weights -> (B, T, 5) one Gaussian (S7)."""
    mu = raw[..., :2]
    sig = F.softplus(raw[..., 2:4]) + 1e-3
    rho = torch.tanh(raw[..., 4])
    w = weights[:, :, None]                                         # (B, M, 1)
    mean = (w[..., None] * mu).sum(1)                               # (B, T, 2)
    d = mu - mean[:, None]
    var = (w[..., None] * (sig ** 2 + d ** 2)).sum(1)
    cov = (w * (rho * sig[..., 0] * sig[..., 1] + d[..., 0] * d[..., 1])).sum(1)
    sx, sy = var[..., 0].sqrt(), var[..., 1].sqrt()
    return torch.stack([mean[..., 0], mean[..., 1], sx, sy, (cov / (sx * sy)).clamp(-0.999, 0.999)], dim=-1)

class Denoiser(nn.Module):
    """Predicts the noise in a noisy residual trajectory, given the encoder context and the step (M3)."""
    def __init__(self, T_f, ctx, n_steps, hidden=256, t_dim=64):
        super().__init__()
        self.t_embed = nn.Embedding(n_steps, t_dim)
        self.net = nn.Sequential(
            nn.Linear(T_f * 2 + t_dim + ctx, hidden), nn.SiLU(),
            nn.Linear(hidden, hidden), nn.SiLU(),
            nn.Linear(hidden, hidden), nn.SiLU(),
            nn.Linear(hidden, T_f * 2))

    def forward(self, x, t, ctx):
        return self.net(torch.cat([x.flatten(1), self.t_embed(t), ctx], dim=-1)).view_as(x)


class C2FTP(nn.Module):
    def __init__(self, input_dim=STATE_DIM, d=64, heads=4, n_modes=6, T_f=TF_STEPS,
                 n_steps=100, ddim_steps=10, n_samples=16, res_scale=5.0):
        super().__init__()
        self.T_f, self.M, self.n_steps = T_f, n_modes, n_steps
        self.ddim_steps, self.n_samples, self.res_scale = ddim_steps, n_samples, res_scale
        self.embed = nn.Linear(input_dim, d)
        self.encoder = nn.LSTM(d, d, batch_first=True)
        self.attn = nn.MultiheadAttention(d, heads, batch_first=True)
        self.norm = nn.LayerNorm(d)
        self.mode_head = nn.Linear(d, n_modes)
        self.traj_head = nn.Sequential(nn.Linear(d, 128), nn.ReLU(), nn.Linear(128, n_modes * T_f * 5))
        self.denoiser = Denoiser(T_f, d, n_steps)
        betas = torch.linspace(1e-4, 0.2, n_steps)                 # M3
        self.register_buffer("alpha_bar", torch.cumprod(1 - betas, dim=0))

    def encode(self, states, mask):
        """Context (B, d) and the coarse single-Gaussian prediction (B, T_f, 5) (M2)."""
        B, T_h, N, _ = states.shape
        _, (h, _) = self.encoder(F.elu(self.embed(states)).transpose(1, 2).reshape(B * N, T_h, -1))
        h = h[-1].reshape(B, N, -1)
        a, _ = self.attn(h[:, :1], h, h, key_padding_mask=~mask)
        context = self.norm(a[:, 0] + h[:, 0])
        raw = self.traj_head(context).reshape(B, self.M, self.T_f, 5)
        return context, moment_match(raw, torch.softmax(self.mode_head(context), dim=-1))

    def aux_loss(self, states, mask, dims, gt):
        """Denoising loss on the residual (M4); called by fit with the true future."""
        context, coarse = self.encode(states, mask)
        r0 = (gt - coarse[..., :2].detach()) / self.res_scale
        t = torch.randint(0, self.n_steps, (gt.shape[0],), device=gt.device)
        noise = torch.randn_like(r0)
        a = self.alpha_bar[t][:, None, None]
        return F.mse_loss(self.denoiser(a.sqrt() * r0 + (1 - a).sqrt() * noise, t, context), noise)

    def refine(self, context):
        """K deterministic DDIM samples of the scaled residual (M6) -> (B, K, T_f, 2)."""
        B, K = context.shape[0], self.n_samples
        g = torch.Generator(device=context.device).manual_seed(SEED)
        x = torch.randn(B * K, self.T_f, 2, generator=g, device=context.device)
        ctx = context.repeat_interleave(K, dim=0)
        steps = torch.linspace(self.n_steps - 1, 0, self.ddim_steps).long().tolist()
        for i, t in enumerate(steps):
            a_t = self.alpha_bar[t]
            a_prev = self.alpha_bar[steps[i + 1]] if i + 1 < len(steps) else torch.ones((), device=x.device)
            eps = self.denoiser(x, torch.full((B * K,), t, device=x.device, dtype=torch.long), ctx)
            x0 = ((x - (1 - a_t).sqrt() * eps) / a_t.sqrt()).clamp(-5, 5)
            x = a_prev.sqrt() * x0 + (1 - a_prev).sqrt() * eps
        return x.reshape(B, K, self.T_f, 2) * self.res_scale

    def forward(self, states, mask, dims=None):
        B, _, N, _ = states.shape
        context, coarse = self.encode(states, mask)
        if not self.training:                                       # M5: refine at evaluation only
            res = self.refine(context)
            mean = coarse[..., :2] + res.mean(dim=1)
            sigma = (coarse[..., 2:4].square() + res.var(dim=1)).sqrt()
            coarse = torch.cat([mean, sigma, coarse[..., 4:]], dim=-1)
        return coarse, states.new_zeros(B, N - 1, 4), None


def build_model():
    return C2FTP()

model = build_model().to(DEVICE)
n_params = sum(p.numel() for p in model.parameters())
print(f"{type(model).__name__}: {n_params:,} parameters")
```
